```python
def SingleBarChartplt(column, label):
    Column = column.value_counts()
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(Column.index, Column.values, color='#1a80bb', edgecolor='black')
    ax.set_title(label)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.show()
    
#SingleDonutplt(df['Smoking'], 'SmokingLevel')
    
    
```

```python
def SingleBarChartPlotly(column, label):
    fig = px.bar(
        x=column.value_counts().index,
        y=column.value_counts().values,
        title=label
    )
    fig.show()
    
#SingleDonutplt(df['Smoking'], 'SmokingLevel')
   
```

```python
def SingleDonutplt(column, label):
    Column = column.value_counts()
    fig, ax = plt.subplots(figsize=(7,7))
    ghach, name = ax.pie(Column.values, labels=Column.index, wedgeprops=dict(width=0.4), startangle=90)
    ax.set_title(label)
    plt.show()
    
    
#SingleDonutplt(df['Smoking'], 'SmokingLevel')
    
    
```

```python
def SingleDonutPlotly(column, label):
    Column = column.value_counts().reset_index()
    fig = px.pie(Column, hole=0.45, title = label)
    fig.show()
    
#SingleDonutplt(df['Smoking'], 'SmokingLevel')
```

```python
def FreqTablePlt(column, label):
    count = column.value_counts()
    Darsad = (count/count.sum() *100)
    fig, ax = plt.subplots(figsize = (7,4))
    ax.axis('off')
    df_display = pd.DataFrame({'Count': count, 'Percent': Darsad})
    Ferekans = ax.table(cellText=df_display.values,  rowLabels=df_display.index, colLabels=df_display.columns, cellLoc='center', loc='center')
    ax.set_title(label)
    
    #Kheyli zesht shode 
    Ferekans.scale(1, 1.5)
    
    plt.show()
    
#FreqTablePlt(df['Smoking'], 'SmokingLevel')
```

```python
def GroupedBarplt(column1, column2, label):
    ct = pd.crosstab(column1, column2)
    ax = ct.plot(kind="bar", figsize=(7,4), width = 0.5)
    plt.title = label
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    plt.show()
    
#GroupedBarplt(df['Smoking'], df['Occupation'], 'Relation of Occupation and Smoking')    
```

```python
def StackedBarplt(column1, column2, label):
    ct = pd.crosstab(column1, column2)
    ax = ct.plot(kind = 'bar', stacked=True, figsize=(7,4))
    plt.title(label)
    plt.show()
    
#StackedBarplt(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking")      
```

```python
def StackedBarplotly(column1, column2, label):
    ct = df.groupby([column1, column2]).size().reset_index(name="idfk")
    fig = px.bar(ct, x=column1.name, y="idfk", color = column2.name,  barmode="stack", title = label)
    
    fig.show()
    
#StackedBarplotly(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking") 
```

```python
def NormalStackedBarplt(column1, column2 , label):
    ct = pd.crosstab(column1, column2,normalize='index') *100
    
    ax = ct.plot(kind = 'bar', stacked=True, figsize=(7,4))
    plt.title(label)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.show()
    
#NormalStackedBarplt(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking")     
```

```python
def NormalStackedBarplotly(column1, column2, label):
    ct = df.groupby([column1, column2]).size().reset_index(name = "Count")
    
    #Used AI for the line below, Sorry :D
    ct["Percent"] = ct.groupby(column1.name)["Count"].transform(lambda x: x / x.sum() * 100)
    
    fig = px.bar(ct, x= column1.name, y= "Percent", color = column2.name, barmode='stack', title= label)
    fig.show()
    
    
#NormalStackedBarplotly(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking")       
```

```python
"""
def Heatmapplt(column1, column2, label):
    
    ct = pd.crosstab(column1, column2)
    
    fig, ax = plt.subplots(figsize = (7,4))
    ax.set_title(label)
"""

def Heatmapplt(column1, column2, label1):
    fig, ax = plt.subplots(figsize=(7,4))
    ct = pd.crosstab(column1, column2)
    im = ax.imshow(ct.values, cmap='Blues')
    
    ax.set_xticks(range(len(ct.columns)))
    ax.set_yticks(range(len(ct.index)))

    ax.set_xticklabels(ct.columns, rotation=45, ha='right')
    ax.set_yticklabels(ct.index)
    
    for i in range(ct.shape[0]):
        for j in range(ct.shape[1]):
            ax.text(j, i, ct.values[i, j],
                    ha='center', va='center', color='black')

    ax.set_title(label1)
    plt.colorbar(im, ax=ax, label='Count')

    plt.show()
    
#Heatmapplt(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking") 
```

```python
def Heatmapplotly(column1, column2, label):
    ct = pd.crosstab(column1, column2)
    
    fig = px.imshow(ct, text_auto = True, color_continuous_scale="YlGnBu", labels= dict(x=column2.name, y=column1.name, color="count"), title=label)
    fig.show()
    
#Heatmapplotly(df['Smoking'], df['Occupation'],"Relation of Occupation and Smoking") 
```
