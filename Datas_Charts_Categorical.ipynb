```python
TwoColumnSummary(df['Binned_Stress Level (1-10)'], df['Binned_Anxiety Level (1-10)'], 'Stress Level vs Anxiety')
TwoColumnSummary(df['Binned_Therapy Sessions (per month)'], df['Binned_Anxiety Level (1-10)'], 'Therapy Sessions vs Anxiety')
TwoColumnSummary(df['Binned_Sleep Hours'], df['Binned_Anxiety Level (1-10)'], 'Sleep vs Anxiety')
TwoColumnSummary(df['Binned_Caffeine Intake (mg/day)'], df['Binned_Anxiety Level (1-10)'], 'Caffeine vs Anxiety')
TwoColumnSummary(df['Binned_Physical Activity (hrs/week)'], df['Binned_Anxiety Level (1-10)'], 'Physical Activity vs Anxiety')
```
