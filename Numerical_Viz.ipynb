```python
# 1 variable -----> Numerical ------> Seaborn
def hist_box_seaborn(data, col):
    fig, axes =  plt.subplots(1, 2, figsize = (10, 4))
    sns.histplot(data = data, x = col, kde=True, color='cornflowerblue', ax= axes[0])
    axes[0].set_title('Histogram Plot')
    

    sns.boxplot(data = data, x = col,ax= axes[1] )
    axes[1].set_title('Scatter Plot')
    axes[1].set_xlabel(col)
    axes[1].set_ylabel('count')
    
```

```python
# Test -----> 1 variable -----> Numerical ------> Seaborn
hist_box_seaborn(data, 'Age')
```

```python
# 1 variable -----> Numerical ------> Pyplot
def hist_box_pyplot(data, col):
    fig = make_subplots(rows = 2, cols= 1)
    fig.add_trace(go.Histogram(x = data[col], name = 'Histogram Plot'), row=1, col=1)
    
    fig.add_trace(go.Box(x = data[col], name = 'Box Plot'), row=2, col=1)
    fig.update_layout(
    
    title =  'Histogram and Boxplot',
    xaxis_title = col,
    yaxis_title = 'Count'
    )

    fig.show()
```

```python
# Test -----> 1 variable -----> Numerical ------> Pyplot
hist_box_pyplot(data, 'Age')
```

```python
# 2 variables -----> Numerical ------> Target ------> Seaborn
def box_violin_target_seaborn(data, target, col):
    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    sns.boxplot(data= data, x = target, y= col,ax= axes[0] )
    axes[0].set_title('Box Plot')
    axes[0].set_xlabel(target)
    axes[0].set_ylabel(col)

    sns.violinplot(data = data, x = data[target], y = data[col], color = 'green', ax=axes[1])
    axes[1].set_title('Violin Plot')
    axes[1].set_xlabel(target)
    axes[1].set_ylabel(col)
```

```python
# Test ---> 2 variables -----> Numerical ------> Target ------> Seaborn
box_violin_target_seaborn(data, target, 'Age')
```

```python
# 2 variables -----> Numerical ------> Target ------> Plotly
def box_violin_target_plotly(data, target, col):
    fig = make_subplots(rows = 2, cols= 1)

    fig.add_trace(go.Box(x = data[target], y=data[col], name = 'Box Plot'), row=1, col=1)

    fig.add_trace(go.Violin(x = data[target], y = data[col], name = 'Violin'), row=2, col=1)
    fig.show()
```

```python
# Test ---> 2 variables -----> Numerical ------> Target ------> Plotly
box_violin_target_plotly(data, target, 'Age')
```

```python
# 2 variables -----> Numerical ------> Target ------> Seaborn
def percentage_target_seaborn(data, target, col):
    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    percentage = data.groupby(target).size().reset_index(name = 'count')

    sns.barplot(data= percentage, x = target, y= 'count',ax= axes[0] )
    axes[0].set_title('Count Plot')
    axes[0].set_xlabel(target)
    axes[0].set_ylabel(col)

    percentage['percentage'] = percentage['count']/ percentage['count'].sum() 

    sns.barplot(data = percentage, x = target, y = 'percentage', color = 'green', ax=axes[1])
    axes[1].set_title('Percentage Plot')
    axes[1].set_xlabel(target)
    axes[1].set_ylabel(col)
```

```python
# Test ---> 2 variables -----> Numerical ------> Target ------> Seaborn
percentage_target_seaborn(data, target, 'Age')
```

```python
# 2 variables -----> Numerical ------> Target ------> Plotly
def percentage_target_plotly(data, target, col):
    fig = make_subplots(rows = 2, cols= 1)

    percentage = data.groupby(target).size().reset_index(name = 'count')

    fig.add_trace(go.Bar(x = percentage[target], y=percentage['count'], name = 'Count Plot'), row=1, col=1)

    percentage['percentage'] = percentage['count']/ percentage['count'].sum() 

    fig.add_trace(go.Bar(x = percentage[target], y = percentage['percentage'], name = 'Percentage Plot'), row=2, col=1)
    fig.show()
```

```python
# Test ---> 2 variables -----> Numerical ------> Target ------> Plotly
percentage_target_plotly(data, target, 'Age')
```

```python
# 2 variables -----> Numerical ------> Seaborn
def heatmap_scatter_seaborn(data, col_1, col_2):
    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    correlation = data[[col_1, col_2]].corr()
    sns.heatmap(correlation, annot= True, cmap= 'viridis', fmt = '.2f', ax = axes[0])
    axes[0].set_title('Heatmap')


    sns.scatterplot(data = data, x = data[col_1], y = data[col_2], ax = axes[1], color = 'green')
    axes[1].set_title('Scatter Plot')
    axes[1].set_xlabel(col_1)
    axes[1].set_ylabel(col_2)
    
```

```python
# Test ----> 2 variables -----> Numerical ------> Seaborn
heatmap_scatter_seaborn(data, 'Age', 'Sleep Hours')
```

```python
# 2 variables -----> Numerical ------> Plotly
def heatmap_scatter_plotly(data, col_1, col_2):
    fig = make_subplots(rows = 2, cols= 1)
    fig.add_trace(go.Histogram2d(x = data[col_1], y=data[col_2], name = 'Heatmap'), row=1, col=1)
    
    fig.add_trace(go.Scatter(x = data[col_1], y = data[col_2], mode='markers', name = 'Scatter Plot'), row=2, col=1)
    fig.update_layout(
    
    title =  'Histogram and Boxplot',
    xaxis_title = col_1,
    yaxis_title = col_2
    )

    fig.show()
```

```python
# Test ----> 2 variables -----> Numerical ------> Plotly
heatmap_scatter_plotly(data, 'Age', 'Sleep Hours')
```

```python
# 2 variables ----> Numerical/Categorical -----> Seaborn
def num_cat_seaborn(data, col_1, col_2):
    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    sns.boxplot(data = data, x = col_1, y= col_2,ax= axes[0] )
    axes[0].set_title('Box Plot')
    axes[0].set_xlabel(col_1)
    axes[0].set_ylabel(col_2)

    sns.violinplot(data = data, x = data[col_1], y = data[col_2], color = 'green', ax=axes[1])
    axes[1].set_title('Violin Plot')
    axes[1].set_xlabel(col_1)
    axes[1].set_ylabel(col_2)
```

```python
# Test -----> 2 variables ----> Numerical/Categorical -----> Seaborn
num_cat_seaborn(data, 'Gender', 'Age')
```

```python
# 2 variables ----> Numerical/Categorical -----> Plotly
def num_cat_plotly(data, col_1, col_2):
    fig = make_subplots(rows = 2, cols= 1)

    fig.add_trace(go.Box(x = data[col_1], y=data[col_2], name = 'Box Plot'), row=1, col=1)

    fig.add_trace(go.Violin(x = data[col_1], y = data[col_2], name = 'Violin'), row=2, col=1)
    fig.show()
```

```python
# Test -----> 2 variables ----> Numerical/Categorical -----> Plotly
num_cat_plotly(data, 'Gender', 'Age')
```

```python
# 3 variables -----> Numerical ------> Seaborn
def three_num_pair_seaborn(data, col_1, col_2, col_3):
  
   three = data[[col_1, col_2, col_3]]
   sns.pairplot(data=three)
```

```python
# Test -----> 3 variables -----> Numerical ------> Seaborn
three_num_pair_seaborn(data, 'Age', 'Sleep Hours', 'Anxiety Level (1-10)')
```

```python
# 3 variables -----> Numerical ------> Seaborn
def three_num_scatter_seaborn(data, col_1, col_2, col_3):
  
   fig = plt.figure(figsize=(12, 4))
   ax = plt.gca()
   sns.scatterplot(data = data, x= data[col_1], y= data[col_2], size= data[col_3], color = 'purple')
   ax.set_xlabel(col_1)
   ax.set_ylabel(col_2)
```

```python
# Test -----> 3 variables -----> Numerical ------> Seaborn
three_num_scatter_seaborn(data, 'Age', 'Sleep Hours', 'Anxiety Level (1-10)')
```

```python
# 3 variables -----> Numerical ------> Plotly
def three_num_plotly(data, col_1, col_2, col_3):
    
    fig= px.scatter(data, x = col_1, y = col_2, size = col_3, title = 'Three columns')
    
    fig.show()
```

```python
# Test -----> 3 variables -----> Numerical ------> Plotly
three_num_plotly(data, 'Age', 'Sleep Hours', 'Anxiety Level (1-10)')
```
