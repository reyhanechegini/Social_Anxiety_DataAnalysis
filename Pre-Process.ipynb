```python
columns = {
    "numeric": [
        "Age",
        "Sleep Hours",
        "Physical Activity (hrs/week)",
        "Caffeine Intake (mg/day)",
        "Alcohol Consumption (drinks/week)",
        "Heart Rate (bpm)",
        "Breathing Rate (breaths/min)",
        "Therapy Sessions (per month)"
    ],

    "ordinal": [
        "Stress Level (1-10)",
        "Sweating Level (1-5)",
        "Diet Quality (1-10)",
        "Anxiety Level (1-10)"
    ],

    "categorical": [
        "Gender",
        "Occupation",
        "Smoking",
        "Family History of Anxiety",
        "Dizziness",
        "Medication",
        "Recent Major Life Event"
    ],

    "target": [
        "Target",
        "is_Anxious"
    ]
}
```

```python
def negative_to_nan(df):
    df = df.copy()

    numeric_cols = df.select_dtypes(include=["number"]).columns

    for col in numeric_cols:
        df.loc[df[col] < 0, col] = np.nan

    return df
```

```python
def outliers_numeric(df, cols):
    df = df.copy()

    for col in cols:

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        df.loc[(df[col] < lower) | (df[col] > upper), col] = np.nan

    return df

#--
```

```python
def clean_ordinal(df, cols):
    df = df.copy()

    ordinal_ranges = {
        "Stress Level (1-10)": (1, 10),
        "Diet Quality (1-10)": (1, 10),
        "Anxiety Level (1-10)": (1, 10),
        "Sweating Level (1-5)": (1, 5)
    }

    for col in cols:
        if col in ordinal_ranges:
            low, high = ordinal_ranges[col]
            df.loc[(df[col] < low) | (df[col] > high), col] = np.nan

    return df
```

```python
def clean_categorical(df, cols):
    df = df.copy()

    for col in cols:
        df[col] = df[col].fillna("Unknown")

    return df

#--
```

```python
df=negative_to_nan(df)
```

```python
df=outliers_numeric(df,columns["numeric"])
df=clean_ordinal(df,columns["ordinal"])
df=clean_categorical(df,columns["categorical"])
```

```python
def fillna_numeric(df, cols):
    df = df.copy()

    for col in cols:

        skew = df[col].skew()

        if abs(skew) < 0.5:
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].median())

    return df
```

```python
def fillna_ordinal(df, cols):
    df = df.copy()

    for col in cols:
        df[col] = df[col].fillna(df[col].median())

    return df
```

```python
df=fillna_numeric(df,columns["numeric"])
df=fillna_ordinal(df,columns["ordinal"])
```

```python
def encode_categorical(df):
    df = df.copy()

    if "Gender" in df.columns:
        df["Gender"] = df["Gender"].map({
            "Male": 0,
            "Female": 1,
            "Unknown": -1
        })

    binary_map = {
        "Yes": 1,
        "No": 0,
        "Unknown": -1
    }

    binary_cols = [
        "Smoking",
        "Family History of Anxiety",
        "Dizziness",
        "Medication",
        "Recent Major Life Event"
    ]

    for col in binary_cols:
        if col in df.columns:
            df[col] = df[col].map(binary_map)

    if "Occupation" in df.columns:
        occupation_map = {
            'Teacher': 0,
            'Student': 1,
            'Lawyer': 2,
            'Nurse': 3,
            'Other': 4,
            'Musician': 5,
            'Engineer': 6,
            'Artist': 7,
            'Chef': 8,
            'Scientist': 9,
            'Athlete': 10,
            'Freelancer': 11,
            'Doctor': 12
        }

        df["Occupation"] = df["Occupation"].map(occupation_map)
    return df
```

```python
df=encode_categorical(df)
```

```python
def bining_features(df):
    df = df.copy()

    bin_config = {

        "Age": (
            [0, 18, 30, 50, 120],
            ["Teen", "Young Adult", "Adult", "Senior"]
        ),

        "Sleep Hours": (
            [0, 5, 7, 9, 24],
            ["Sleep Deprived", "Normal Sleep", "Healthy Sleep", "Oversleeping"]
        ),

        "Physical Activity (hrs/week)": (
            [-1, 2, 5, 100],
            ["Non-Athletic", "Moderately Active", "Athletic"]
        ),

        "Caffeine Intake (mg/day)": (
            [-1, 100, 300, 600, 2000],
            ["Low", "Moderate", "High", "Coffee Addict"]
        ),

        "Alcohol Consumption (drinks/week)": (
            [-1, 0, 5, 10, 100],
            ["None", "Light", "Moderate", "Heavy"]
        ),

        "Heart Rate (bpm)": (
            [0, 60, 100, 200],
            ["Low", "Normal", "High"]
        ),

        "Breathing Rate (breaths/min)": (
            [0, 12, 20, 100],
            ["Slow", "Normal", "Rapid"]
        ),

        "Therapy Sessions (per month)": (
            [-1, 0, 2, 5, 20],
            ["No Therapy", "Occasional", "Regular", "Intensive"]
        ),

        "Stress Level (1-10)": (
            [0, 3, 6, 10],
            ["Low Stress", "Moderate Stress", "High Stress"]
        ),

        "Diet Quality (1-10)": (
            [0, 3, 6, 8, 10],
            ["Unhealthy", "Average", "Healthy", "Excellent"]
        ),

        "Anxiety Level (1-10)": (
            [0, 3, 6, 10],
            ["Low Anxiety", "Moderate Anxiety", "High Anxiety"]
        )

    }

    for col in bin_config:

        bins, labels = bin_config[col]

        df["Binned_",col ] = pd.cut(
            df[col],
            bins=bins,
            labels=labels
        )
    return df

 
```

```python
df=bining_features(df)
```
