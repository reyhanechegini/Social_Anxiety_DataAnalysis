```python
columns = {
    "numeric": [
        "Age",
        "Sleep Hours",
        "Physical Activity (hrs/week)",
        "Caffeine Intake (mg/day)",
        "Alcohol Consumption (drinks/week)",
        "Heart Rate (bpm)",
        "Breathing Rate (breaths/min)",
        "Therapy Sessions (per month)"
    ],

    "ordinal": [
        "Stress Level (1-10)",
        "Sweating Level (1-5)",
        "Diet Quality (1-10)",
        "Anxiety Level (1-10)"
    ],

    "categorical": [
        "Gender",
        "Occupation",
        "Smoking",
        "Family History of Anxiety",
        "Dizziness",
        "Medication",
        "Recent Major Life Event"
    ],

    "target": [
        "Target",
        "is_Anxious"
    ]
}
```

```python
limits= {

    "Age": (0, 100),

    "Sleep Hours": (0, 24),

    "Physical Activity (hrs/week)": (0, 70),

    "Caffeine Intake (mg/day)": (0, 1000),

    "Alcohol Consumption (drinks/week)": (0, 30),

    "Stress Level (1-10)": (1, 10),

    "Heart Rate (bpm)": (40, 200),

    "Breathing Rate (breaths/min)": (8, 40),

    "Sweating Level (1-5)": (1, 5),

    "Therapy Sessions (per month)": (0, 10),

    "Diet Quality (1-10)": (1, 10),

    "Anxiety Level (1-10)": (1, 10)
}
```

```python
def fix_therapy_history(df):
    df = df.copy()

    
    mask = (df["Therapy Sessions (per month)"] > 0) & (df["Therapy History"].isna())

    df.loc[mask, "Therapy History"] = "Psychodynamic Therapy"

   
    df.loc[df["Therapy Sessions (per month)"] == 0, "Therapy History"] = np.nan

    

    return df
```

```python
def clean_outliers(df, limits):
    df = df.copy()

    for col, (low, high) in limits.items():

        if col in df.columns:
            df[col] = df[col].clip(lower=low, upper=high)

    return df
```

```python
def clean_categorical(df, cols):
    df = df.copy()

    for col in cols:

        mode = df[col].mode()

        if not mode.empty:
            df[col] = df[col].fillna(mode[0])

    return df
```

```python
df=clean_outliers(df,limits)
df=clean_categorical(df,columns["categorical"])
```

```python
def fillna_numeric(df, cols):
    df = df.copy()

    for col in cols:

        df[col + "_missing"] = df[col].isna().astype(int)

        skew = df[col].skew()

        if abs(skew) < 0.5:
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].median())

    return df
```

```python
def fillna_ordinal(df, cols):
    df = df.copy()

    for col in cols:

        df[col + "_missing"] = df[col].isna().astype(int)

        df[col] = df[col].fillna(df[col].median())

    return df
```

```python
df=fillna_numeric(df,columns["numeric"])
df=fillna_ordinal(df,columns["ordinal"])
df=fix_therapy_history(df)
```

```python
def encode_categorical(df):
    df = df.copy()

    if "Gender" in df.columns:
        df["Gender"] = df["Gender"].map({
            "Male": 0,
            "Female": 1
        })

    binary_map = {
        "Yes": 1,
        "No": 0
    }

    binary_cols = [
        "Smoking",
        "Family History of Anxiety",
        "Dizziness",
        "Medication",
        "Recent Major Life Event"
    ]

    for col in binary_cols:
        if col in df.columns:
            df[col] = df[col].map(binary_map)

    
    # Occupation Reclassification
   
    occupation_reclass = {

        "Teacher": "Academic",
        "Student": "Academic",
        "Scientist": "Academic",

        "Doctor": "Medical",
        "Nurse": "Medical",

        "Engineer": "Tech",
        "Freelancer": "Tech",

        "Artist": "Creative",
        "Musician": "Creative",

        "Lawyer": "Other",
        "Chef": "Other",
        "Athlete": "Other",
        "Other": "Other"
    }

    occupation_encode = {
        "Academic": 0,
        "Medical": 1,
        "Tech": 2,
        "Creative": 3,
        "Other": 4
    }

    if "Occupation" in df.columns:
        df["Occupation"] = df["Occupation"].map(occupation_reclass)
        df["Occupation"] = df["Occupation"].map(occupation_encode)

    return df
```

```python
df=encode_categorical(df)
```

```python
def bin_features(df):
    df = df.copy()

    df["Age Group"] = pd.cut(
        df["Age"],
        bins=[0, 18, 30, 50, 120],
        labels=["Teen", "Young Adult", "Adult", "Senior"]
    )

    df["Sleep Group"] = pd.cut(
        df["Sleep Hours"],
        bins=[0, 5, 7, 9, 24],
        labels=["Very Low", "Low", "Normal", "High"]
    )

    df["Activity Group"] = pd.cut(
        df["Physical Activity (hrs/week)"],
        bins=[-1, 2, 5, 100],
        labels=["Sedentary", "Moderate", "Active"]
    )

    df["Caffeine Group"] = pd.cut(
        df["Caffeine Intake (mg/day)"],
        bins=[-1, 100, 300, 600, 2000],
        labels=["Low", "Moderate", "High", "Very High"]
    )

    df["Alcohol Group"] = pd.cut(
        df["Alcohol Consumption (drinks/week)"],
        bins=[-1, 0, 5, 10, 100],
        labels=["None", "Light", "Moderate", "Heavy"]
    )

    df["Heart Rate Group"] = pd.cut(
        df["Heart Rate (bpm)"],
        bins=[0, 60, 100, 200],
        labels=["Low", "Normal", "High"]
    )

    df["Breathing Rate Group"] = pd.cut(
        df["Breathing Rate (breaths/min)"],
        bins=[0, 12, 20, 100],
        labels=["Slow", "Normal", "Rapid"]
    )

    df["Therapy Group"] = pd.cut(
        df["Therapy Sessions (per month)"],
        bins=[-1, 0, 2, 5, 20],
        labels=["None", "Occasional", "Regular", "Frequent"]
    )

    df["Stress Group"] = pd.cut(
        df["Stress Level (1-10)"],
        bins=[0, 3, 6, 10],
        labels=["Low", "Medium", "High"]
    )

    df["Anxiety Group"] = pd.cut(
        df["Anxiety Level (1-10)"],
        bins=[0, 3, 6, 10],
        labels=["Low", "Medium", "High"]
    )

    df["Diet Group"] = pd.cut(
        df["Diet Quality (1-10)"],
        bins=[0, 3, 6, 8, 10],
        labels=["Unhealthy", "Average", "Healthy", "Excellent"]
    )

    return df

 
```

```python
df=bin_features(df)
```
